```python
import getpass
import json
from openai import OpenAI

api_key = getpass.getpass('Enter your OPENAI_API_KEY: ')
client = OpenAI(api_key=api_key)
```

```python
def solve_exercise(text):
    prompt = f"""You are an expert English teacher. Solve this exercise:
    {text}
    Return JSON: {{'question': '...', 'answer': '...', 'confidence': 0.9, 'explanation': '...'}}"""
    
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[{"role": "user", "content": prompt}],
        response_format={"type": "json_object"}
    )
    print(json.dumps(json.loads(response.choices[0].message.content), indent=2))
```

```python
exercise_text = """
Paste your exercise text here.
Example: Complete the sentence. She (go/goes) to the store.
"""
solve_exercise(exercise_text)
```
